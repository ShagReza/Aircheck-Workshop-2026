```python
print("### Basic Operations ###")

a = 15
b = 4

print("Addition (a + b):", a + b)
print("Subtraction (a - b):", a - b)
print("Multiplication (a * b):", a * b)
print("Division (a / b):", a / b)
print("Integer Division (a // b):", a // b)
print("Modulus (a % b):", a % b)
print("Exponentiation (a ** b):", a ** b)
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Control Flow (if, for, while)</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">Control flow statements in Python allow you to control the execution of code blocks. These include:</p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;"><strong>Conditional statements (if/else)</strong> to make decisions.</li><li style="margin:4px 0;"><strong>For loops</strong> to iterate over a sequence.</li><li style="margin:4px 0;"><strong>While loops</strong> to repeat a block of code as long as a condition is true.</li></ul><p style="margin:7px 0;">We will practice these concepts with examples to understand how Python handles decision-making and repetition.</p></div>

```python
# ✅ Control Flow (if, for, while)
print("### Control Flow (if, for, while) ###")

# If-Else
num = 7
if num % 2 == 0:
    print(f"{num} is Even")
else:
    print(f"{num} is Odd")

# For loop
print("\nFor Loop Example:")
for i in range(1, 6):
    print("Square of", i, "is", i ** 2)

# While loop
print("\nWhile Loop Example:")
count = 0
while count < 3:
    print("Count is:", count)
    count += 1
```

```python
print("### Functions ###")

def greet(name):
    """Function to greet a person"""
    return f"Hello, {name}! Welcome to Python."

def add_numbers(a, b):
    """Function to add two numbers"""
    return a + b

# Using functions
print(greet("Data Scientist"))
print("Sum of 10 and 20:", add_numbers(10, 20))

```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Lists, Tuples, and Dictionaries</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;"><strong>Lists</strong> are ordered and mutable collections of items.</li><li style="margin:4px 0;"><strong>Tuples</strong> are ordered and immutable collections of items.</li><li style="margin:4px 0;"><strong>Dictionaries</strong> are collections of key-value pairs.</li></ul><p style="margin:7px 0;">Here, we will learn how to create and access these data structures in Python, and when to use each of them.</p></div>

```python

print("### Lists, Tuples, and Dictionaries ###")

# List (mutable)
fruits = ["apple", "banana", "cherry"]
print("List of fruits:", fruits)
print("First fruit:", fruits[0])
fruits.append("orange")
print("List after adding orange:", fruits)

# Tuple (immutable)
coordinates = (10, 20)
print("\nTuple of coordinates:", coordinates)
print("X coordinate:", coordinates[0])

# Dictionary (key-value pairs)
person = {"name": "Alice", "age": 30, "city": "New York"}
print("\nDictionary of person:", person)
print("Person's name:", person["name"])

# Adding a new key-value pair
person["job"] = "Engineer"
print("Updated dictionary:", person)
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">NumPy Basics</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">NumPy (Numerical Python) is a fundamental package for scientific computing in Python. It provides support for arrays, matrices, and many mathematical functions.<br>In this section, we will learn how to create NumPy arrays and perform basic operations on them.</p></div>

```python
import numpy as np

print("### NumPy Basics ###")
# Creating 1D array
arr = np.array([1, 2, 3, 4, 5])
print("1D Array:", arr)

# Creating 2D array (matrix)
matrix = np.array([[1, 2, 3], [4, 5, 7]])
print("\n2D Array (Matrix):\n", matrix)

# Checking data type and shape
print("\nArray Data Type:", arr.dtype)
print("Matrix Shape:", matrix.shape)

# Basic statistics
print("Sum of Array:", arr.sum())
print("Mean of Array:", arr.mean())
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Array Operations</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">NumPy arrays allow element-wise operations, which makes computation efficient and concise.<br>We will explore arithmetic operations, broadcasting, and other ways to manipulate arrays in this part.</p></div>

```python
print("\n### Array Operations ###")
# Element-wise arithmetic operations
print("Add 10 to each element:", arr + 10)
print("Square of each element:", arr ** 2)
print("Multiply by 3:", arr * 3)

# Matrix multiplication (dot product)
a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [1, 3]])
print("\nMatrix A:\n", a)
print("Matrix B:\n", b)
print("Matrix Multiplication (A * B):\n", np.dot(a, b))
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Array Indexing and Slicing</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">Just like lists in Python, NumPy arrays can be indexed and sliced to access specific elements or subarrays.<br>We will learn how to retrieve specific elements, rows, and columns from NumPy arrays.</p></div>

```python
print("\n### Array Indexing and Slicing ###")
# Indexing
print("First element of array:", arr[0])
print("Last element of array:", arr[-1])

# Slicing
print("First three elements:", arr[:3])
print("Elements from index 2 onwards:", arr[2:])
print("Every second element:", arr[::2])

# Indexing in 2D array
print("\nElement at (0,1):", matrix[0, 1])
print("First row:", matrix[0, :])
print("Second column:", matrix[:, 1])
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Useful NumPy Functions</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">NumPy offers many built-in functions for creating arrays, generating random numbers, and performing mathematical operations.<br>In this section, we will explore functions like <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">zeros</code>, <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">ones</code>, <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">arange</code>, and <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">random</code>.</p></div>

```python
print("\n### Useful NumPy Functions ###")
# Array of zeros
zeros_array = np.zeros((2, 3))
print("Array of Zeros:\n", zeros_array)

# Array of ones
ones_array = np.ones((3, 3))
print("\nArray of Ones:\n", ones_array)

# Array with a range of values
range_array = np.arange(0, 10, 2)
print("\nArray with range (0 to 10 with step 2):", range_array)

# Random array
random_array = np.random.rand(2, 4)  # 2x4 matrix with random values between 0 and 1
print("\nRandom Array (2x4):\n", random_array)

# Random integers
rand_int_array = np.random.randint(1, 100, size=(3, 3))
print("\nRandom Integers (3x3, 1 to 100):\n", rand_int_array)
```

<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">Selecting specific rows, columns, or both is essential in data analysis.<br>We will learn various ways to filter and select data, including boolean indexing and label-based selection.</p></div>

```python
# Data Selection and Filtering with Pandas (Healthcare Example)

import pandas as pd

print("### Data Selection and Filtering - Healthcare Dataset ###\n")

# Sample healthcare-related DataFrame
data = {
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [45, 62, 37, 29, 54],
    "Gender": ["Female", "Male", "Female", "Male", "Female"],
    "Diagnosis": ["Diabetes", "Hypertension", "Healthy", "Asthma", "Diabetes"],
    "Treatment_Cost": [5000, 7000, 1500, 3000, 4500]
}

df = pd.DataFrame(data)
print("Sample Healthcare DataFrame:\n", df)

print("\n--- Selecting Columns ---")

# Selecting a single column (Series)
print("\nSingle Column (Diagnosis):\n", df["Diagnosis"])

# Selecting multiple columns (DataFrame)
print("\nMultiple Columns (Patient_ID and Treatment_Cost):\n", df[["Patient_ID", "Treatment_Cost"]])


print("\n--- Selecting Rows ---")

# Selecting rows using iloc (index-based selection)
print("\nRow at index 2 (using iloc):\n", df.iloc[2])


# Selecting rows using loc (label-based selection, here labels are indices)
print("\nRow with Patient_ID 104 (using loc):\n", df.loc[df["Patient_ID"] == 104])

print("\n--- Selecting Rows and Columns Together ---")

# Selecting specific rows and columns

print("\nSelecting 'Age' and 'Diagnosis' for first 3 patients:\n", df.loc[0:2, ["Age", "Diagnosis"]])

print("\n--- Conditional Filtering (Boolean Indexing) ---")

# Filtering rows based on a condition (e.g., Age > 50)
print("\nPatients older than 50:\n", df[df["Age"] > 50])

# Filtering with multiple conditions (e.g., Female patients with Diabetes)
print("\nFemale patients diagnosed with Diabetes:\n", df[(df["Gender"] == "Female") & (df["Diagnosis"] == "Diabetes")])

# Filtering rows where Treatment Cost > 4000
print("\nPatients with Treatment Cost over 4000:\n", df[df["Treatment_Cost"] > 4000])

print("\n--- Important Notes ---")
print("- Use single brackets df['column'] to select a Series (one column).")
print("- Use double brackets df[['col1', 'col2']] to select multiple columns (DataFrame).")
print("- Use df.loc[row_label, column_label] for label-based selection or conditional selection.")
print("- Use df.iloc[row_index, column_index] for index-based selection.")
print("- Boolean indexing is essential for filtering healthcare data based on diagnosis, age, cost, and other attributes.")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Basic Operations on DataFrames</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">Pandas provides various functions to perform operations on dataframes, such as:</p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;">Summarizing data with <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">describe()</code></li><li style="margin:4px 0;">Adding and removing columns</li><li style="margin:4px 0;">Sorting and aggregating data</li></ul><p style="margin:7px 0;">In this section, we will explore these essential functions.</p></div>

```python
# Basic Operations on DataFrames (Healthcare Dataset)

import pandas as pd

print("### Basic Operations on DataFrames - Healthcare Dataset ###\n")

# Sample healthcare dataset
data = {
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [45, 62, 37, 29, 54],
    "Gender": ["Female", "Male", "Female", "Male", "Female"],
    "Diagnosis": ["Diabetes", "Hypertension", "Healthy", "Asthma", "Diabetes"],
    "Treatment_Cost": [5000, 7000, 1500, 3000, 4500]
}

df = pd.DataFrame(data)
print("Healthcare DataFrame:\n", df)

print("\n--- Summarizing Data ---")

# Summarize numerical columns
print("\nSummary statistics (describe):\n", df.describe())

# Summarize categorical columns
print("\nSummary of non-numeric data:\n", df.describe(include=['object']))

print("\n--- Adding a New Column ---")

# Add a new column 'Hospital_Stay_Days'
df["Hospital_Stay_Days"] = [5, 7, 2, 3, 4]
print("\nDataFrame with new column 'Hospital_Stay_Days':\n", df)

print("\n--- Removing a Column ---")

# Remove the 'Hospital_Stay_Days' column
df.drop(columns=["Hospital_Stay_Days"], inplace=True)
print("\nDataFrame after removing 'Hospital_Stay_Days':\n", df)

print("\n--- Sorting Data ---")

# Sort patients by Age
print("\nSorted by Age (ascending):\n", df.sort_values(by="Age"))

# Sort by Treatment Cost (descending order)
print("\nSorted by Treatment_Cost (descending):\n", df.sort_values(by="Treatment_Cost", ascending=False))

print("\n--- Aggregating Data ---")

# Aggregate: Average Treatment Cost by Diagnosis
avg_cost_by_diagnosis = df.groupby("Diagnosis")["Treatment_Cost"].mean()
print("\nAverage Treatment Cost by Diagnosis:\n", avg_cost_by_diagnosis)

# Aggregate: Count of patients by Diagnosis
patient_count_by_diagnosis = df["Diagnosis"].value_counts()
print("\nNumber of patients by Diagnosis:\n", patient_count_by_diagnosis)

print("\n--- Important Notes ---")
print("- 'describe()' provides statistical summary for numeric columns.")
print("- New columns can be added by direct assignment (e.g., df['New_Column'] = values).")
print("- Use 'drop()' to remove columns or rows (set 'inplace=True' to apply changes directly).")
print("- Sorting can be done ascending or descending using 'sort_values()'.")
print("- Aggregations like mean(), sum(), count() can be performed using 'groupby' or 'value_counts'.")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Handling Missing Data</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">Real-world datasets often contain missing data. Pandas provides several ways to handle these issues, including:</p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;">Checking for missing data</li><li style="margin:4px 0;">Filling missing values with specific values or computed ones</li><li style="margin:4px 0;">Dropping rows or columns with missing data</li></ul><p style="margin:7px 0;">We will explore these techniques to clean our data for analysis.</p></div>

```python
# Handling Missing Data in Pandas (Healthcare Dataset)

import pandas as pd
import numpy as np

print("### Handling Missing Data - Healthcare Dataset ###\n")

# Sample healthcare dataset with missing values
data = {
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [45, np.nan, 37, 29, 54],                  # Missing age for Patient 102
    "Gender": ["Female", "Male", np.nan, "Male", "Female"],  # Missing gender for Patient 103
    "Diagnosis": ["Diabetes", "Hypertension", "Healthy", "Asthma", "Diabetes"],
    "Treatment_Cost": [5000, 7000, np.nan, 3000, 4500]  # Missing cost for Patient 103
}

df = pd.DataFrame(data)
print("Healthcare DataFrame with Missing Values:\n", df)

print("\n--- Identifying Missing Data ---")

# Check for missing values in each column
print("\nMissing Values (per column):\n", df.isnull().sum())

# Total number of missing values
print("\nTotal number of missing values in DataFrame:", df.isnull().sum().sum())

print("\n--- Filling Missing Data ---")

# Filling missing numerical data with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())
print("\nFilled missing 'Age' with mean:\n", df['Age'])

# Filling missing categorical data with mode (most frequent value)
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
print("\nFilled missing 'Gender' with mode:\n", df['Gender'])

# Filling missing 'Treatment_Cost' with median (good for skewed data)
df['Treatment_Cost'] = df['Treatment_Cost'].fillna(df['Treatment_Cost'].median())
print("\nFilled missing 'Treatment_Cost' with median:\n", df['Treatment_Cost'])

print("\nDataFrame after filling missing values:\n", df)

print("\n--- Removing Rows/Columns with Missing Data (Example) ---")

# (Recreate the original DataFrame to demonstrate dropna)
df_with_na = pd.DataFrame(data)  # With missing values again for demo

# Drop rows with any missing values
dropped_rows = df_with_na.dropna()
print("\nDataFrame after dropping rows with any missing values:\n", dropped_rows)

# Drop columns with any missing values (for larger datasets where columns may be removed)
dropped_columns = df_with_na.dropna(axis=1)
print("\nDataFrame after dropping columns with missing values:\n", dropped_columns)

print("\n--- Important Notes ---")
print("- Use 'isnull()' to detect missing values.")
print("- Use 'fillna()' to fill missing data (mean, median, mode, custom values).")
print("- Use 'dropna()' to remove rows or columns with missing values.")
print("- Filling strategies depend on data type and context: mean/median for numbers, mode for categories.")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Applying Functions to the Entire DataFrame</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;">Using <strong>apply() and map()</strong> to modify DataFrame columns efficiently.</li><li style="margin:4px 0;">Applying <strong>lambda functions</strong> to perform custom transformations.</li><li style="margin:4px 0;">Demonstrating <strong>vectorized operations</strong> for performance optimization.</li></ul><p style="margin:7px 0;">These techniques are essential when working with healthcare data to:</p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;">Categorize patients (e.g., age groups).</li><li style="margin:4px 0;">Encode categorical data (e.g., gender mapping).</li><li style="margin:4px 0;">Adjust costs based on patient attributes.</li><li style="margin:4px 0;">Apply bulk transformations efficiently.</li></ul></div>

```python
import pandas as pd

print("### Applying Functions to DataFrame - Healthcare Dataset ###\n")

# Sample healthcare dataset
data = {
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [45, 62, 37, 29, 54],
    "Gender": ["Female", "Male", "Female", "Male", "Female"],
    "Diagnosis": ["Diabetes", "Hypertension", "Healthy", "Asthma", "Diabetes"],
    "Treatment_Cost": [5000, 7000, 1500, 3000, 4500]
}

df = pd.DataFrame(data)
print("Original Healthcare DataFrame:\n", df)

print("\n--- Using apply() to modify columns ---")

# Example: Apply a function to categorize patients as "Senior" if Age > 50
df['Age_Group'] = df['Age'].apply(lambda x: 'Senior' if x > 50 else 'Adult')
print("\nAdded 'Age_Group' column based on Age:\n", df[['Patient_ID', 'Age', 'Age_Group']])

print("\n--- Using map() to transform categorical data ---")

# Example: Mapping Gender to numeric format (Female=0, Male=1)
gender_map = {'Female': 0, 'Male': 1}
df['Gender_Code'] = df['Gender'].map(gender_map)
print("\nMapped 'Gender' to 'Gender_Code':\n", df[['Patient_ID', 'Gender', 'Gender_Code']])

print("\n--- Using lambda to modify multiple columns ---")

# Example: Increase Treatment Cost by 10% for patients over 50
df['Adjusted_Cost'] = df.apply(lambda row: row['Treatment_Cost'] * 1.10 if row['Age'] > 50 else row['Treatment_Cost'], axis=1)
print("\nAdjusted Treatment Cost (10% increase if Age > 50):\n", df[['Patient_ID', 'Age', 'Treatment_Cost', 'Adjusted_Cost']])

print("\n--- Vectorized Operations (Efficient Way) ---")

# Example: Apply discount of $500 to all Treatment Costs efficiently
df['Discounted_Cost'] = df['Treatment_Cost'] - 500
print("\nDiscounted Treatment Cost (Vectorized operation):\n", df[['Patient_ID', 'Treatment_Cost', 'Discounted_Cost']])

print("\n--- Important Notes ---")
print("- 'apply()' can apply functions row-wise or column-wise (use axis=1 for rows).")
print("- 'map()' is used for element-wise transformations, especially useful for categorical data.")
print("- Lambda functions allow creating quick inline functions without defining them explicitly.")
print("- Vectorized operations are the fastest way to perform element-wise operations on entire columns (recommended when possible).")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Merging &amp; Combining DataFrames</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">In real-world healthcare data, information often comes from <strong>multiple sources</strong> (e.g., patient records, treatments, diagnostics).</p><p style="margin:7px 0;">In this section, we will learn:</p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;">How to <strong>combine datasets</strong> using <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">merge()</code>, <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">concat()</code>, and <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">join()</code>.</li><li style="margin:4px 0;">The difference between <strong>inner, outer, left, and right joins</strong>.</li><li style="margin:4px 0;">How to handle datasets when patients may or may not appear in both sources.</li></ul><p style="margin:7px 0;">These techniques are essential for building <strong>complete and accurate</strong> datasets for analysis and reporting.</p></div>

```python
import pandas as pd

print("### Merging & Combining DataFrames - Healthcare Dataset ###\n")

# Sample patient information DataFrame
df_patients = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [45, 62, 37, 29, 54],
    "Gender": ["Female", "Male", "Female", "Male", "Female"]
})

# Sample diagnosis and treatment DataFrame
df_treatments = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 106],  # Notice Patient_ID 106 is only here
    "Diagnosis": ["Diabetes", "Hypertension", "Healthy", "Asthma"],
    "Treatment_Cost": [5000, 7000, 1500, 3500]
})

# Additional DataFrame for concatenation
df_followup = pd.DataFrame({
    "Patient_ID": [107, 108],
    "Name": ["Frank", "Grace"],
    "Age": [50, 41],
    "Gender": ["Male", "Female"]
})

print("Patient DataFrame:\n", df_patients)
print("\nTreatment DataFrame:\n", df_treatments)

print("\n--- Merging DataFrames with merge() ---")

# Inner Join: Only patients who appear in both DataFrames
inner_merged = pd.merge(df_patients, df_treatments, on='Patient_ID', how='inner')
print("\nInner Join (common patients only):\n", inner_merged)

# Outer Join: All patients from both DataFrames, missing values filled with NaN
outer_merged = pd.merge(df_patients, df_treatments, on='Patient_ID', how='outer')
print("\nOuter Join (all patients from both):\n", outer_merged)

# Left Join: All patients from df_patients, matched data from df_treatments
left_merged = pd.merge(df_patients, df_treatments, on='Patient_ID', how='left')
print("\nLeft Join (all patients, even if no treatment):\n", left_merged)

# Right Join: All patients from df_treatments, matched data from df_patients
right_merged = pd.merge(df_patients, df_treatments, on='Patient_ID', how='right')
print("\nRight Join (all treated patients, even if not in patient list):\n", right_merged)

print("\n--- Combining DataFrames with concat() ---")

# Concatenating patients and follow-up patients (vertical stack)
combined_patients = pd.concat([df_patients, df_followup], ignore_index=True)
print("\nConcatenated Patients (original + follow-up):\n", combined_patients)

print("\n--- Joining DataFrames with join() ---")

# Example with DataFrames sharing the same index (must set index first)
df_left = df_patients.set_index('Patient_ID')
df_right = df_treatments.set_index('Patient_ID')

# Join: Left join on index
joined_df = df_left.join(df_right, how='left')
print("\nJoined DataFrame (based on Patient_ID index):\n", joined_df)

print("\n--- Important Notes ---")
print("- 'merge()' is used for relational joins, similar to SQL joins.")
print("- 'concat()' stacks DataFrames vertically or horizontally.")
print("- 'join()' is handy when joining on index, not columns.")
print("- Be mindful of 'how' parameter: inner, outer, left, right.")
print("- Handling real-world data often requires merging from multiple sources, especially in healthcare datasets like patient records and treatments.")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Detecting &amp; Handling Duplicates</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;">Identifying duplicate rows.</li><li style="margin:4px 0;">Removing unnecessary duplicates.</li><li style="margin:4px 0;">Cleaning inconsistencies in real-world datasets.</li></ul></div>

```python
import pandas as pd

print("### Detecting & Handling Duplicates - Healthcare Dataset ###\n")

# Sample healthcare dataset with intentional duplicate rows
data = {
    "Patient_ID": [101, 102, 103, 104, 105, 102, 105],  # Duplicate IDs 102, 105
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Bob", "Eva"],  # Duplicate names matching Patient_IDs
    "Age": [45, 62, 37, 29, 54, 62, 54],  # Consistent with Patient_ID duplicates
    "Gender": ["Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "Diagnosis": ["Diabetes", "Hypertension", "Healthy", "Asthma", "Diabetes", "Hypertension", "Diabetes"],
    "Treatment_Cost": [5000, 7000, 1500, 3000, 4500, 7000, 4500]
}

# Create DataFrame
df = pd.DataFrame(data)
print("Healthcare DataFrame with Duplicates:\n", df)

print("\n--- Identifying Duplicates ---")

# Detect all duplicate rows
duplicates = df.duplicated(keep=False)  # Shows all duplicates (keep=False keeps all instances)
print("\nDuplicate Rows (True indicates duplicate):\n", duplicates)

# Detect duplicates considering all columns
duplicate_rows = df[df.duplicated()]
print("\nDuplicate Rows (First occurrence kept, subsequent marked duplicate):\n", duplicate_rows)

print("\n--- Removing Duplicates ---")

# Remove exact duplicate rows, keeping the first occurrence
df_no_duplicates = df.drop_duplicates()
print("\nDataFrame after removing exact duplicates:\n", df_no_duplicates)

# Remove duplicates based on a specific column (e.g., Patient_ID), keeping the first occurrence
df_unique_patients = df.drop_duplicates(subset=['Patient_ID'], keep='first')
print("\nDataFrame with unique Patient_ID (first occurrence kept):\n", df_unique_patients)

print("\n--- Handling Inconsistencies ---")

# Example: Detect inconsistent entries (e.g., same Patient_ID but different Diagnosis)
inconsistent_patients = df.groupby('Patient_ID').agg({'Diagnosis': pd.Series.nunique})
inconsistent_patients = inconsistent_patients[inconsistent_patients['Diagnosis'] > 1]
print("\nPatients with inconsistent Diagnoses (more than one unique diagnosis):\n", inconsistent_patients)

print("\n--- Important Notes ---")
print("- 'duplicated()' identifies duplicate rows. Use 'keep=False' to mark all duplicates.")
print("- 'drop_duplicates()' removes duplicate rows; 'subset' can limit check to specific columns.")
print("- 'groupby' with 'nunique' can help detect inconsistencies in important columns like diagnosis for the same patient.")
print("- In real-world healthcare datasets, cleaning duplicates ensures accurate patient records and analyses.")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Correlation &amp; Feature Relationships</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">In this section, we will:</p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;"><strong>Compute correlations</strong> between numerical columns to identify relationships.</li><li style="margin:4px 0;"><strong>Visualize correlations</strong> using a heatmap for better interpretation.</li><li style="margin:4px 0;"><strong>Discuss implications</strong> for feature engineering in machine learning.</li></ul></div>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

print("### Correlation & Feature Relationships - Healthcare Dataset ###\n")

# Sample healthcare dataset
data = {
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [45, 62, 37, 29, 54],
    "Treatment_Cost": [5000, 7000, 1500, 3000, 4500],
    "Hospital_Stay_Days": [5, 7, 2, 3, 4]
}

# Create DataFrame
df = pd.DataFrame(data)
print("Healthcare DataFrame:\n", df)

print("\n--- Computing Correlations ---")

# Correlation matrix for numerical columns
correlation_matrix = df.corr()
print("\nCorrelation Matrix:\n", correlation_matrix)

print("\n--- Visualizing Correlation Matrix with Heatmap ---")

# Heatmap to visualize correlations
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, linewidths=0.5, linecolor='black', square=True)
plt.title('Correlation Heatmap: Numerical Features in Healthcare Data')
plt.show()

print("\n--- Insights for Feature Engineering ---")
print("- High correlation between features may indicate redundancy.")
print("- Weak correlations can highlight unrelated features, which may not be useful for predictive models.")
print("- Understanding feature relationships helps in selecting relevant variables for machine learning models.")
print("- For example, if 'Age' and 'Treatment_Cost' are highly correlated, they may represent similar underlying patterns.")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Removing Highly Correlated Features Before ML Modeling</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">Highly correlated features can cause redundancy in machine learning models, leading to overfitting and poor generalization. In this section, we will:</p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;"><strong>Compute correlations</strong> and identify highly correlated pairs.</li><li style="margin:4px 0;"><strong>Remove one of the features</strong> from each correlated pair to simplify the feature set.</li><li style="margin:4px 0;">Prepare a <strong>cleaner dataset</strong> for machine learning.</li></ul><p style="margin:7px 0;">This step is critical in feature engineering and improves model robustness.</p></div>

```python
import pandas as pd
import numpy as np  # Correct way to import NumPy

print("### Removing Highly Correlated Features Before ML Modeling ###\n")

# Sample healthcare dataset
data = {
    "Patient_ID": [101, 102, 103, 104, 105],
    "Age": [45, 62, 37, 29, 54],
    "Treatment_Cost": [5000, 7000, 1500, 3000, 4500],
    "Hospital_Stay_Days": [5, 7, 2, 3, 4]
}

# Create DataFrame
df = pd.DataFrame(data)

# Drop Patient_ID since it's an identifier, not a feature
df_features = df.drop(columns=['Patient_ID'])

print("Original Features:\n", df_features)

print("\n--- Computing Correlation Matrix ---")
correlation_matrix = df_features.corr().abs()  # Get absolute correlation values
print("\nAbsolute Correlation Matrix:\n", correlation_matrix)
print("\n--- Identifying Highly Correlated Features ---")

# Select upper triangle of correlation matrix (to avoid duplicate checks)
upper = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

# Set correlation threshold (e.g., 0.8)
threshold = 0.8

# Find columns with correlation higher than threshold
to_drop = [column for column in upper.columns if any(upper[column] > threshold)]
print(f"\nFeatures to drop (correlation > {threshold}): {to_drop}")

# Drop highly correlated columns
df_reduced = df_features.drop(columns=to_drop)
print("\nFeatures after removing highly correlated columns:\n", df_reduced)

print("\n--- Important Notes ---")
print("- Highly correlated features may provide redundant information to ML models.")
print("- Removing them helps reduce overfitting and improves model generalization.")
print("- Correlation threshold (e.g., 0.8) can be adjusted as needed.")
print("- Always combine statistical analysis with domain expertise when selecting features.")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">A Larger Sample Dataset for Plotting</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">Real insight needs real volume. Here we generate <strong>200 synthetic patients</strong> with NumPy's random<br>number generator. The columns are deliberately built with relationships baked in — longer hospital<br>stays cost more, older patients stay longer — so the plots and correlations later have something<br>genuine to reveal.</p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;"><code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">np.random.default_rng(seed)</code> is the modern NumPy random generator.</li><li style="margin:4px 0;">The <strong>seed</strong> makes the data reproducible: everyone in the room gets identical numbers.</li><li style="margin:4px 0;"><code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">rng.normal()</code> draws from a bell curve, <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">rng.choice()</code> picks categories with given probabilities.</li></ul></div>
<div style="background:#fffbeb;border-left:5px solid #f59e0b;border-radius:0 10px 10px 0;padding:14px 20px;margin:14px 0;color:#7c2d12;line-height:1.7;font-size:14.5px;"><div style="font-weight:800;color:#b45309;margin-bottom:6px;">&#128204; Note</div><p style="margin:7px 0;">This data is <strong>synthetic</strong> — invented for teaching. Never present conclusions from simulated data<br>as clinical findings. What transfers to real work is the <em>method</em>, not the numbers.</p></div>

```python
# Build a larger synthetic healthcare dataset (200 patients)

import numpy as np
import pandas as pd

rng = np.random.default_rng(42)   # fixed seed -> everyone gets the same numbers
n = 200

# --- Demographics --------------------------------------------------
age = rng.normal(loc=55, scale=15, size=n).clip(18, 95).round(0)
gender = rng.choice(["Female", "Male"], size=n, p=[0.52, 0.48])
bmi = rng.normal(loc=27, scale=4.5, size=n).clip(15, 50).round(1)

diagnosis = rng.choice(
    ["Diabetes", "Hypertension", "Asthma", "Healthy"],
    size=n,
    p=[0.30, 0.30, 0.15, 0.25],
)

# --- Clinical outcomes ---------------------------------------------
# Built so the columns carry a REAL relationship we can rediscover later.
severity = {"Diabetes": 2.5, "Hypertension": 2.0, "Asthma": 1.5, "Healthy": 0.0}

# Older and sicker patients tend to stay longer...
stay = age / 20 + np.array([severity[d] for d in diagnosis]) + rng.normal(0, 1.0, n)
stay = stay.clip(1, None).round(0)

# ...and cost is driven mostly by length of stay, a little by age.
cost = 900 + 620 * stay + 25 * age + rng.normal(0, 400, n)

# A small share of patients suffer complications, which push their bill far
# above everyone else. This is what gives real healthcare cost data its
# long right tail - and it is what we will detect as skew and outliers later.
complication = rng.random(n) < 0.08
cost = cost + complication * rng.gamma(shape=2.0, scale=2600, size=n)
cost = cost.clip(300, None).round(2)

df_big = pd.DataFrame({
    "Patient_ID": np.arange(1001, 1001 + n),
    "Age": age.astype(int),
    "Gender": gender,
    "BMI": bmi,
    "Diagnosis": diagnosis,
    "Hospital_Stay_Days": stay.astype(int),
    "Treatment_Cost": cost,
})

print("Shape (rows, columns):", df_big.shape)

print("\nFirst 5 rows:")
print(df_big.head())

print("\nNumeric summary:")
print(df_big.describe().round(2))

print("\nPatients per diagnosis:")
print(df_big["Diagnosis"].value_counts())
```

We use the<br>object-oriented style throughout, because it scales to multi-panel figures without surprises.</p></div>
<div style="background:#fffbeb;border-left:5px solid #f59e0b;border-radius:0 10px 10px 0;padding:14px 20px;margin:14px 0;color:#7c2d12;line-height:1.7;font-size:14.5px;"><div style="font-weight:800;color:#b45309;margin-bottom:6px;">&#128204; Note</div><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;">In Jupyter and Colab, figures display automatically — <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">plt.show()</code> simply marks the end of one figure.</li><li style="margin:4px 0;">Call <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">fig.savefig(...)</code> <strong>before</strong> <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">plt.show()</code>; after showing, the figure is cleared and you save a blank file.</li><li style="margin:4px 0;"><code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">figsize=(width, height)</code> is in <strong>inches</strong>, and <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">dpi</code> controls resolution — <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">dpi=150</code> is a good default for slides.</li></ul></div>

```python
# Matplotlib fundamentals: one Figure, one Axes
# (requires df_big from the section above)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)

fig, ax = plt.subplots(figsize=(8, 4.5))    # fig = the canvas, ax = one plot on it

ax.plot(x, np.sin(x), label="sin(x)", color="#0d9488", linewidth=2)
ax.plot(x, np.cos(x), label="cos(x)", color="#f59e0b", linewidth=2, linestyle="--")

ax.set_title("Your first Matplotlib figure", fontsize=14, fontweight="bold")
ax.set_xlabel("x")
ax.set_ylabel("f(x)")
ax.set_xlim(0, 10)
ax.legend(loc="upper right")
ax.grid(alpha=0.3)

plt.show()


# ----------------------------------------------------------------------
# Several plots on one canvas: plt.subplots(rows, cols) returns a grid of Axes
# ----------------------------------------------------------------------
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
axes = axes.ravel()          # flatten the 2x2 grid into a list of 4 Axes

# (a) Histogram - the shape of one numeric column
axes[0].hist(df_big["Age"], bins=20, color="#0d9488", edgecolor="white")
axes[0].set_title("Histogram: patient age")
axes[0].set_xlabel("Age (years)")
axes[0].set_ylabel("Number of patients")

# (b) Scatter - the relationship between two numeric columns
axes[1].scatter(df_big["Hospital_Stay_Days"], df_big["Treatment_Cost"],
                s=20, alpha=0.55, color="#0f766e")
axes[1].set_title("Scatter: stay length vs cost")
axes[1].set_xlabel("Hospital stay (days)")
axes[1].set_ylabel("Treatment cost (USD)")

# (c) Horizontal bar - comparing a summary across categories
mean_cost = df_big.groupby("Diagnosis")["Treatment_Cost"].mean().sort_values()
axes[2].barh(mean_cost.index, mean_cost.values, color="#14b8a6")
axes[2].set_title("Bar: mean cost by diagnosis")
axes[2].set_xlabel("Mean treatment cost (USD)")

# (d) Line - a trend across ordered groups
age_band = pd.cut(df_big["Age"], bins=[18, 30, 40, 50, 60, 70, 95])
trend = df_big.groupby(age_band, observed=True)["Treatment_Cost"].mean()
axes[3].plot([str(i) for i in trend.index], trend.values,
             marker="o", color="#f59e0b", linewidth=2)
axes[3].set_title("Line: mean cost by age band")
axes[3].set_ylabel("Mean cost (USD)")
axes[3].tick_params(axis="x", rotation=45)

fig.suptitle("Four plot types, one figure", fontsize=15, fontweight="bold")
fig.tight_layout()                                     # stop the panels overlapping
fig.savefig("healthcare_overview.png", dpi=150, bbox_inches="tight")   # save BEFORE show
plt.show()

print("Figure saved to 'healthcare_overview.png'")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Histograms, Boxplots, and Reading a Distribution</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;"><code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">describe()</code> gives you the numbers; a <strong>histogram</strong> gives you the shape. Both matter — very<br>different distributions can share the same mean and standard deviation.</p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;">A <strong>histogram</strong> splits a numeric column into <em>bins</em> and counts how many values fall in each.</li><li style="margin:4px 0;">The <strong>bin count is a choice, not a fact</strong>: too few bins hide structure, too many turn signal into noise.</li><li style="margin:4px 0;"><strong>Skew</strong> means the distribution leans to one side. When data is skewed, the <em>mean</em> is dragged toward the long tail and the <em>median</em> is the safer summary — this is exactly why we filled missing <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">Treatment_Cost</code> with the median earlier.</li><li style="margin:4px 0;">A <strong>boxplot</strong> compresses a distribution into median, interquartile range (IQR), whiskers, and outlier points — which makes it ideal for comparing several groups side by side.</li></ul></div>
<div style="background:#fffbeb;border-left:5px solid #f59e0b;border-radius:0 10px 10px 0;padding:14px 20px;margin:14px 0;color:#7c2d12;line-height:1.7;font-size:14.5px;"><div style="font-weight:800;color:#b45309;margin-bottom:6px;">&#128204; Note</div><p style="margin:7px 0;">The common outlier rule is <strong>1.5 × IQR</strong> below the first quartile or above the third.<br>It is a convention, not a law: in healthcare data an "outlier" cost is often a real, expensive<br>patient rather than a data-entry error. Investigate before you delete.</p></div>

```python
# Histograms, skew, and boxplots
# (requires df_big from the "Larger Sample Dataset" section)

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ----------------------------------------------------------------------
# 1. The same data, three bin counts - the bin choice tells a different story
# ----------------------------------------------------------------------
fig, axes = plt.subplots(1, 3, figsize=(13, 3.8), sharey=True)

for ax, bins in zip(axes, [5, 15, 50]):
    ax.hist(df_big["Age"], bins=bins, color="#0d9488", edgecolor="white")
    ax.set_title(f"bins = {bins}")
    ax.set_xlabel("Age (years)")

axes[0].set_ylabel("Number of patients")
fig.suptitle("Same column, three bin counts", fontsize=14, fontweight="bold")
fig.tight_layout()
plt.show()


# ----------------------------------------------------------------------
# 2. Skew: where mean and median disagree
# ----------------------------------------------------------------------
cost = df_big["Treatment_Cost"]

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.hist(cost, bins=25, color="#99f6e4", edgecolor="#0f766e")
ax.axvline(cost.mean(), color="#b91c1c", linestyle="--", linewidth=2,
           label=f"mean   = {cost.mean():,.0f}")
ax.axvline(cost.median(), color="#1d4ed8", linestyle="-", linewidth=2,
           label=f"median = {cost.median():,.0f}")
ax.set_title("Distribution of treatment cost", fontsize=14, fontweight="bold")
ax.set_xlabel("Treatment cost (USD)")
ax.set_ylabel("Number of patients")
ax.legend()
plt.show()

print(f"Skewness of Treatment_Cost: {cost.skew():.2f}")
print("  > 0  -> long tail to the right (a few very expensive patients)")
print("  ~ 0  -> roughly symmetric")
print("  < 0  -> long tail to the left")


# ----------------------------------------------------------------------
# 3. Boxplots: comparing distributions across groups
# ----------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(9, 4.5))
sns.boxplot(data=df_big, x="Diagnosis", y="Treatment_Cost",
            hue="Diagnosis", palette="crest", legend=False, ax=ax)
ax.set_title("Treatment cost by diagnosis", fontsize=14, fontweight="bold")
ax.set_ylabel("Treatment cost (USD)")
plt.show()


# ----------------------------------------------------------------------
# 4. Counting outliers with the 1.5 x IQR rule
# ----------------------------------------------------------------------
q1, q3 = cost.quantile([0.25, 0.75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
outliers = cost[(cost < lower) | (cost > upper)]

print(f"\nQ1 = {q1:,.0f}   Q3 = {q3:,.0f}   IQR = {iqr:,.0f}")
print(f"Normal range: {lower:,.0f} to {upper:,.0f}")
print(f"Outliers flagged: {len(outliers)} of {len(cost)} patients")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Scatter Plots and Correlation, Done Properly</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">A <strong>scatter plot</strong> shows how two numeric columns move together; <strong>correlation</strong> puts a single number<br>on that relationship. Together they are the fastest way to spot which features carry real signal —<br>and which ones are duplicates of each other.</p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;"><strong>Pearson</strong> (<code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">method="pearson"</code>, the default) measures a <em>straight-line</em> relationship.</li><li style="margin:4px 0;"><strong>Spearman</strong> (<code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">method="spearman"</code>) ranks the values first, so it catches any relationship that is consistently increasing or decreasing, even if it curves.</li><li style="margin:4px 0;"><strong>Watch the gap between the two.</strong> For stay length vs cost you will get Pearson ≈ 0.51 but Spearman ≈ 0.74. The handful of complication patients from the previous section drag the straight-line fit off course, while the rank-based measure shrugs them off. A big Pearson-Spearman gap is a signal that outliers or a curve are in play.</li><li style="margin:4px 0;">A <strong>heatmap</strong> of the correlation matrix reads faster than the table. Masking the upper triangle removes the mirrored duplicates so your eye goes to the real content.</li><li style="margin:4px 0;">A <strong>pairplot</strong> draws every scatter combination at once — the best single command for a first look.</li></ul></div>
<div style="background:#fffbeb;border-left:5px solid #f59e0b;border-radius:0 10px 10px 0;padding:14px 20px;margin:14px 0;color:#7c2d12;line-height:1.7;font-size:14.5px;"><div style="font-weight:800;color:#b45309;margin-bottom:6px;">&#128204; Note: Two traps to remember</div><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;"><strong>Correlation is not causation.</strong> Cost and stay length move together here because we built them that way; in real data, a third hidden variable is often driving both.</li><li style="margin:4px 0;"><strong>r ≈ 0 does not mean "no relationship".</strong> Pearson only sees straight lines. A perfect U-shaped curve scores r ≈ 0 — the last cell below proves it. Always plot before you trust the number.</li></ul></div>

```python
# Scatter plots and correlation on a realistic dataset
# (requires df_big from the "Larger Sample Dataset" section)

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = ["Age", "BMI", "Hospital_Stay_Days", "Treatment_Cost"]

# ----------------------------------------------------------------------
# 1. Scatter plots with a fitted trend line
# ----------------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

for ax, xcol in zip(axes, ["Age", "Hospital_Stay_Days"]):
    x = df_big[xcol]
    y = df_big["Treatment_Cost"]

    ax.scatter(x, y, s=20, alpha=0.55, color="#0f766e")

    # np.polyfit(x, y, 1) fits a straight line and returns [slope, intercept]
    slope, intercept = np.polyfit(x, y, 1)
    xs = np.linspace(x.min(), x.max(), 50)
    ax.plot(xs, slope * xs + intercept, color="#b45309", linewidth=2, label="trend line")

    ax.set_title(f"{xcol} vs Treatment_Cost   (r = {x.corr(y):.2f})", fontweight="bold")
    ax.set_xlabel(xcol.replace("_", " "))
    ax.set_ylabel("Treatment cost (USD)")
    ax.legend()

fig.tight_layout()
plt.show()


# ----------------------------------------------------------------------
# 2. Pearson vs Spearman
# ----------------------------------------------------------------------
print("Pearson correlation (straight-line relationships):")
print(df_big[num_cols].corr(method="pearson").round(2))

print("\nSpearman correlation (rank-based, catches curved trends):")
print(df_big[num_cols].corr(method="spearman").round(2))


# ----------------------------------------------------------------------
# 3. Heatmap with the mirrored half masked out
# ----------------------------------------------------------------------
corr = df_big[num_cols].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))    # hide the duplicated upper triangle

fig, ax = plt.subplots(figsize=(6.5, 5))
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="coolwarm",
            center=0, vmin=-1, vmax=1, square=True, linewidths=0.5,
            cbar_kws={"shrink": 0.8}, ax=ax)
ax.set_title("Correlation heatmap (numeric features)", fontsize=13, fontweight="bold")
plt.show()


# ----------------------------------------------------------------------
# 4. Pairplot: every scatter combination in one command
# ----------------------------------------------------------------------
grid = sns.pairplot(df_big[num_cols + ["Diagnosis"]], hue="Diagnosis",
                    corner=True, height=1.9, plot_kws={"s": 16, "alpha": 0.6})
grid.figure.suptitle("Pairwise relationships by diagnosis", y=1.02,
                     fontsize=13, fontweight="bold")
plt.show()


# ----------------------------------------------------------------------
# 5. The trap: a perfect relationship that Pearson scores as zero
# ----------------------------------------------------------------------
x_curve = np.linspace(-3, 3, 200)
y_curve = x_curve ** 2                       # y is perfectly determined by x

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.scatter(x_curve, y_curve, s=12, color="#b91c1c")
ax.set_title(f"y = x²   ->   Pearson r = {np.corrcoef(x_curve, y_curve)[0, 1]:.2f}",
             fontweight="bold")
plt.show()

print("A perfect curved relationship scores r = 0.00 with Pearson.")
print("Correlation near zero means 'no STRAIGHT-LINE trend', not 'no relationship'. Always plot.")
```

<div style="background:#ecfeff;border-left:5px solid #0d9488;border-radius:0 8px 8px 0;padding:10px 18px;margin:22px 0 0 0;"><span style="color:#0f766e;font-size:17px;font-weight:700;">Putting It All Together: A Mini EDA Challenge</span></div>
<div style="background:#f0fdfa;border-left:5px solid #0d9488;border-radius:0 10px 10px 0;padding:14px 20px;margin:0 0 8px 0;color:#134e4a;line-height:1.7;font-size:14.5px;"><p style="margin:7px 0;">Exploratory Data Analysis (EDA) is the same handful of questions asked over and over. The function<br>below packages them into one reusable call — the kind of helper worth keeping in your own toolkit.</p><p style="margin:7px 0;"><strong>Run it, read it, then take on the tasks at the bottom of the cell.</strong></p><ul style="margin:6px 0;padding-left:20px;"><li style="margin:4px 0;">What does one numeric column look like on its own? <em>(histogram)</em></li><li style="margin:4px 0;">How does it differ across groups? <em>(boxplot)</em></li><li style="margin:4px 0;">Which group is biggest, and which is most expensive? <em>(count bar + mean bar)</em></li><li style="margin:4px 0;">What do the numbers say to back up the picture? <em>(<code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">groupby().agg()</code>)</em></li></ul></div>
<div style="background:#fffbeb;border-left:5px solid #f59e0b;border-radius:0 10px 10px 0;padding:14px 20px;margin:14px 0;color:#7c2d12;line-height:1.7;font-size:14.5px;"><div style="font-weight:800;color:#b45309;margin-bottom:6px;">&#128204; Tip</div><p style="margin:7px 0;">Write plotting helpers that <strong>return</strong> the figure instead of only calling <code style="background:#ccfbf1;color:#134e4a;padding:1px 6px;border-radius:4px;font-size:0.92em;">plt.show()</code>. That way the<br>same function works in a notebook, in a script that saves PNGs, and inside a report generator.</p></div>

```python
# A reusable four-panel EDA helper
# (requires df_big from the "Larger Sample Dataset" section)

import matplotlib.pyplot as plt
import seaborn as sns


def quick_eda(df, num_col, cat_col, palette="crest"):
    """Four-panel snapshot of one numeric column, split by one categorical column.

    Parameters
    ----------
    df      : pandas DataFrame
    num_col : name of the numeric column to profile
    cat_col : name of the categorical column to group by
    palette : any Seaborn palette name

    Returns
    -------
    The Matplotlib Figure, so the caller can save or further customise it.
    """
    fig, axes = plt.subplots(2, 2, figsize=(11.5, 7))
    axes = axes.ravel()

    # (a) Overall distribution
    axes[0].hist(df[num_col], bins=20, color="#0d9488", edgecolor="white")
    axes[0].axvline(df[num_col].median(), color="#b91c1c", linestyle="--",
                    linewidth=2, label="median")
    axes[0].set_title(f"Distribution of {num_col}")
    axes[0].set_xlabel(num_col)
    axes[0].set_ylabel("Count")
    axes[0].legend()

    # (b) Distribution per group
    sns.boxplot(data=df, x=cat_col, y=num_col, hue=cat_col,
                palette=palette, legend=False, ax=axes[1])
    axes[1].set_title(f"{num_col} by {cat_col}")
    axes[1].tick_params(axis="x", rotation=20)

    # (c) How many rows in each group
    counts = df[cat_col].value_counts()
    axes[2].bar(counts.index, counts.values, color="#14b8a6")
    axes[2].set_title(f"Rows per {cat_col}")
    axes[2].set_ylabel("Count")
    axes[2].tick_params(axis="x", rotation=20)

    # (d) Group averages, ranked
    means = df.groupby(cat_col)[num_col].mean().sort_values()
    axes[3].barh(means.index, means.values, color="#f59e0b")
    axes[3].set_title(f"Mean {num_col} by {cat_col}")
    axes[3].set_xlabel(f"Mean {num_col}")

    fig.suptitle(f"EDA snapshot - {num_col} across {cat_col}",
                 fontsize=15, fontweight="bold")
    fig.tight_layout()
    return fig


# --- Run it -----------------------------------------------------------
quick_eda(df_big, "Treatment_Cost", "Diagnosis")
plt.show()

# The numbers behind the picture
summary = (df_big
           .groupby("Diagnosis")["Treatment_Cost"]
           .agg(patients="count", mean="mean", median="median", std="std")
           .round(1)
           .sort_values("mean", ascending=False))
print("Treatment cost by diagnosis:\n")
print(summary)


# ======================================================================
# YOUR TURN - uncomment one line at a time and read what the plots say
# ======================================================================

# TODO 1: Profile hospital stay across genders instead.
# quick_eda(df_big, "Hospital_Stay_Days", "Gender"); plt.show()

# TODO 2: Which diagnosis group has the widest spread of BMI?
# quick_eda(df_big, "BMI", "Diagnosis"); plt.show()

# TODO 3: Add an "Age_Group" column, then profile cost across it.
#         Hint: pd.cut(df_big["Age"], bins=[18, 40, 60, 95],
#                      labels=["Young", "Middle", "Senior"])

# TODO 4: Save one of your figures at presentation quality.
#         Hint: fig = quick_eda(...); fig.savefig("my_eda.png", dpi=150, bbox_inches="tight")

# TODO 5: Which two numeric columns are the most correlated, and does the
#         scatter plot agree with the correlation number?
#         Hint: df_big[["Age", "BMI", "Hospital_Stay_Days", "Treatment_Cost"]].corr()
```
